```python
# Ask the user to enter a temperature in Celsius
celsius = float(input("Enter temperature in °C: "))

# Convert Celsius to Fahrenheit
fahrenheit = (celsius * 9/5) + 32

# Display the converted temperature
print(f"{celsius}°C is equal to {fahrenheit:.2f}°F")
```

```python
# Set the temperature threshold
threshold = 30

# Check whether the temperature is above or below the threshold
if celsius > threshold:
    print(f"The temperature is above the threshold of {threshold}°C.")
elif celsius < threshold:
    print(f"The temperature is below the threshold of {threshold}°C.")
else:
    print(f"The temperature is exactly at the threshold of {threshold}°C.")
```

```python
# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

# Test the function
temperature = 25
converted_temperature = celsius_to_fahrenheit(temperature)

print(f"{temperature}°C is equal to {converted_temperature:.2f}°F")
```

```python
# A list of temperatures to convert
temperatures = [0, 10, 20, 30, 40]

# Convert and display each temperature
for temperature in temperatures:
    fahrenheit = celsius_to_fahrenheit(temperature)
    print(f"{temperature}°C = {fahrenheit:.2f}°F")
```
