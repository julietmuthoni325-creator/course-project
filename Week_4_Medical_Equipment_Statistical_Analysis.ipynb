# Statistical Analysis of Equipment Failure Data

## Project Problem
Medical equipment can develop faults unexpectedly. My project explores how equipment operating conditions and usage data can help identify patterns associated with equipment failure.

## Objective
The objective of this analysis is to calculate descriptive statistics, examine the relationship between numerical variables and understand how data can support future equipment failure prediction.

## Dataset
This analysis uses the AI4I 2020 Predictive Maintenance Dataset. It contains simulated industrial machine data, not hospital medical equipment data, so it is used as a relevant example of predictive maintenance.

```python

# Import the pandas library
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/ai4i2020.csv')

# Display the first five rows
df.head()
```

```python

# Select four numerical variables relevant to equipment condition
variables = [
    'Air temperature [K]',
    'Process temperature [K]',
    'Rotational speed [rpm]',
    'Torque [Nm]'
]

# Display the selected variables
df[variables].head()
```

```python

# Calculate descriptive statistics for each variable

statistics = pd.DataFrame({
    'Mean': df[variables].mean(),
    'Median': df[variables].median(),
    'Range': df[variables].max() - df[variables].min(),
    'Variance': df[variables].var(),
    'Standard Deviation': df[variables].std()
})

# Display the results rounded to two decimal places
statistics.round(2)
```

```python

# Select two variables for comparison
x = df['Air temperature [K]']
y = df['Process temperature [K]']

# Calculate covariance and Pearson correlation
covariance = x.cov(y)
correlation = x.corr(y)

print("Covariance:", round(covariance, 4))
print("Correlation:", round(correlation, 4))
```

```python

# Import matplotlib for plotting
import matplotlib.pyplot as plt

# Create a scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(x, y, alpha=0.5)

plt.xlabel('Air Temperature (K)')
plt.ylabel('Process Temperature (K)')
plt.title('Relationship Between Air and Process Temperature')
plt.grid(True)
plt.show()
```


## Interpretation of Results

### What Do the Statistics Tell Us?
The mean and median describe the typical values of the variables. The range shows the difference between the highest and lowest values, while variance and standard deviation indicate how much the values vary. These statistics help us understand the operating conditions represented in the dataset.

### What Does the Correlation Tell Us?
The correlation between air temperature and process temperature indicates the direction and strength of their linear relationship. I will use the calculated correlation value to determine whether the relationship is positive or negative and whether it is weak or strong.

### Does Correlation Mean Causation?
No. Correlation does not necessarily mean that one variable causes the other. Two variables may be related because of other factors or shared operating conditions. Further investigation is needed to establish causation.
